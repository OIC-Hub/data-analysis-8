```python
# GENERATE THE DATASET
import numpy as np

np.random.seed(42)
```

```python
consumption = np.random.randint(200, 601, size=120)

print("DAILY ELECTRICITY CONSUMPTION")
print(consumption)
```

```python
# BASIC STATISTICS
mean_consumption = np.mean(consumption)
median_consumption = np.median(consumption)
minimum_consumption = np.min(consumption)
maximum_consumption = np.max(consumption)
std_consumption = np.std(consumption)
```

```python
print("\nBASIC STATISTICS")
print("Mean:", mean_consumption, "kWh")
print("Median:", median_consumption, "kWh")
print("Minimum:", minimum_consumption, "kWh")
print("Maximum:", maximum_consumption, "kWh")
print("Standard Deviation:", std_consumption, "kWh")
```

```python
# IDENTIFY EXTREMES
highest_days = np.where(consumption == maximum_consumption)[0] + 1
lowest_days = np.where(consumption == minimum_consumption)[0] + 1
```

```python
print("\nIDENTIFY EXTREMES")
print("Highest consumption:", maximum_consumption, "kWh")
print("Day(s) with highest consumption:", highest_days)
print("Lowest consumption:", minimum_consumption, "kWh")
print("Day(s) with lowest consumption:", lowest_days)
```

```python
# MONTHLY SPLITTING
month1 = consumption[0:30]
month2 = consumption[30:60]
month3 = consumption[60:90]
month4 = consumption[90:120]
```

```python
print("\nMONTHLY DATA")
print("Month 1:", month1)
print("Month 2:", month2)
print("Month 3:", month3)
print("Month 4:", month4)
```

```python
# MONTHLY ANALYSIS
months = [month1, month2, month3, month4]
```

```python
print("\nMONTHLY ANALYSIS")

for i, month in enumerate(months, start=1):
    total = np.sum(month)
    average = np.mean(month)

    print(f"Month {i}:")
    print("Total consumption:", total, "kWh")
    print("Average daily consumption:", round(average, 2), "kWh")
```

```python
# 6. WEEKLY ANALYSIS
# There are 17 complete weeks in 120 days.
# The remaining 1 day is ignored.
```

```python
complete_days = consumption[:(len(consumption) // 7) * 7]
weeks = complete_days.reshape(-1, 7)

weekly_totals = np.sum(weeks, axis=1)
weekly_averages = np.mean(weeks, axis=1)

print("\nWEEKLY ANALYSIS")
```

```python
for i in range(len(weeks)):
    print(f"Week {i + 1}:")
    print("Weekly total:", weekly_totals[i], "kWh")
    print("Weekly average:", round(weekly_averages[i], 2), "kWh")
```

```python
# 7. ANOMALY DETECTION
upper_limit = mean_consumption + (2 * std_consumption)
lower_limit = mean_consumption - (2 * std_consumption)
```

```python
anomaly_indices = np.where(
    (consumption > upper_limit) |
    (consumption < lower_limit)
)[0]

anomaly_days = anomaly_indices + 1

print("\nANOMALY DETECTION")
print("Upper limit:", round(upper_limit, 2), "kWh")
print("Lower limit:", round(lower_limit, 2), "kWh")
print("Anomaly day numbers:", anomaly_days)
print("Anomaly consumption values:", consumption[anomaly_indices])
```
