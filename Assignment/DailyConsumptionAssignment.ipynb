```python
import numpy as np
import pandas as pd
```

```python
days = np.arange(1, 121)
consumption = np.random.randint(200, 601, size=120)
```

```python
print(days)
print(consumption)
```

```python
data = pd.DataFrame({
    "Days": days,
    "Consumption": consumption 
})

data
```

```python
#basic statistics
mean_consumption = np.mean(consumption)
std_consumption = np.std(consumption)
print("Average consumption:", mean_consumption)
print("Median consumption:", np.median(consumption))
print("Minimum consumption:", np.min(consumption ))
print("Maximum consumption:", np.max(consumption))
print("Standard deviation:", np.std(consumption))
```

```python
highest = np.max(consumption)
lowest = np.min(consumption)
HighestDays = days[consumption == highest ]
LowestDays = days[consumption == lowest ]
print(f"Highest consumption: {highest} kwh")
print(f"Days with highest consumption: {HighestDays}") 
print(f"Lowest consumption: {lowest} kwh")
print(f"Days with lowest consumption: {LowestDays}")
```

```python
#monthly analysis (split into 4 months )
Months = np.array_split(consumption, 4)
print("Monthly total analysis:")
month_total = [print(month.sum()) for month in Months]
print("\nMonthly average analysis:")
month_avg = [print(month.mean()) for month in Months]
```

```python
#Weekly analysis (split into 7 days per week )
weeklist = (consumption [ :119])
weeks = np.array_split(weeklist, 17)
print("Weekly total analysis:")
Week_total = [print(week.sum()) for week in weeks]
print("\nWeekly average analysis:")
week_avg = [print(week.mean()) for week in weeks]
```

```python
#anomaly detection
lower_bound = mean_consumption - 2 * std_consumption 
upper_bound =  mean_consumption + 2 * std_consumption 

print(lower_bound)
print(upper_bound)

anomaly = np.where((consumption < lower_bound) | (consumption > upper_bound))[0] + 1
print("Anomaly detection (low/high):", anomaly)
```
